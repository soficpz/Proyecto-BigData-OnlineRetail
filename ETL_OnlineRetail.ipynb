```python
df = df.dropna(subset=["Customer ID"])
```

```python
print("Filas después:", len(df))
```

```python
df["Customer ID"].isnull().sum()
```

```python
df = df.drop_duplicates()
df.duplicated().sum()
print("Filas finales:", len(df))
```

```python
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"]) # Transformacion y dimension del tiempo
df["Año"] = df["InvoiceDate"].dt.year
df["Mes"] = df["InvoiceDate"].dt.month
df["Dia"] = df["InvoiceDate"].dt.day
```

```python
df[["InvoiceDate","Año","Mes","Dia"]].head()
```

```python
df["TotalVenta"] = df["Quantity"] * df["Price"] #Crear TotalVenta
df[["Quantity","Price","TotalVenta"]].head()
```

```python
df.to_csv("OnlineRetail_Limpio.csv", index=False)
```
